```python
# [Step 1-1] Civitai API 토큰 설정 (Google Drive .env 자동 확인/저장)
import os

bridge_dir = "/content/drive/MyDrive/AI_GameDev_Bridge"
os.makedirs(bridge_dir, exist_ok=True)
env_file = os.path.join(bridge_dir, ".env")

if not os.path.exists(env_file):
    token = input("Civitai API 토큰을 붙여넣으세요 (이미 저장되어 있으면 엔터): ").strip()
    if token:
        with open(env_file, "w", encoding="utf-8") as f:
            f.write(f"CIVITAI_TOKEN={token}\n")
        print("✓ Civitai 토큰이 Google Drive에 안전하게 저장되었습니다.")
    else:
        print("토큰 입력 없이 계속 진행합니다.")
else:
    print("✓ Google Drive에 Civitai 인증 파일(.env)이 이미 존재합니다.")
```
